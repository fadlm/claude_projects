## Step 1: Anatomy of an Artificial Neuron (Linear Combination)\nAn artificial neuron computes: $z = \sum w_i x_i + b$.

```python
import numpy as np
x = np.array([0.80, 0.60, 0.90])
w = np.array([0.90, 0.70, 1.00])
b = 0.12
z = np.dot(w, x) + b
print(f"Accumulated score (z): {z:.4f}")
```

## Step 2: Non-Linear Activation Functions\nSigmoid: $\sigma(z) = \frac{1}{1 + e^{-z}}$, ReLU: $\text{ReLU}(z) = \max(0, z)$.

```python
def sigmoid(val): return 1.0 / (1.0 + np.exp(-val))
def relu(val): return np.maximum(0.0, val)
print(f"Sigmoid: {sigmoid(z):.4f} | ReLU: {relu(z):.4f}")
```

## Step 3: Measuring Error with Squared Loss\n$\mathcal{L} = (y - \hat{y})^2$.

```python
y_true, y_pred = 1.0, 0.88
loss = (y_true - y_pred) ** 2
print(f"Loss: {loss:.4f}")
```

## Step 4: 4-Step Training Loop\nGuess -> Measure -> Blame -> Adjust.

```python
heights = np.array([1.60, 1.70, 1.80, 1.90])
labels = np.array([0.0, 0.0, 1.0, 1.0])
w, b, lr = 0.10, 0.00, 0.50
for epoch in range(1000):
  preds = sigmoid(w * heights + b)
  dL_dz = -2.0 * (labels - preds) * (preds * (1.0 - preds))
w -= lr * np.mean(dL_dz * heights)
b -= lr * np.mean(dL_dz)
print(f"Trained w: {w:.3f}, b: {b:.3f}")
```

## Step 5: Visual Data Representation & Partitioning\nNormalized 8x8 handwritten digits split 80% train / 20% test.

```python
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
digits = load_digits()
X_scaled = digits.data / 16.0
X_train, X_test, y_train, y_test = train_test_split(X_scaled, digits.target, test_size=0.20, random_state=42, stratify=digits.target)
print(f"Train size: {X_train.shape[0]} | Test size: {X_test.shape[0]}")
```

## Step 6: Multi-Layer Neural Network with PyTorch\n64 pixels -> Linear(64, 32) -> ReLU -> Linear(32, 16) -> ReLU -> Linear(16, 10).

```python
import torch
import torch.nn as nn
import torch.optim as optim
t_X_train, t_y_train = torch.tensor(X_train, dtype=torch.float32), torch.tensor(y_train, dtype=torch.long)
t_X_test, t_y_test = torch.tensor(X_test, dtype=torch.float32), torch.tensor(y_test, dtype=torch.long)
model = nn.Sequential(nn.Linear(64, 32), nn.ReLU(), nn.Linear(32, 16), nn.ReLU(), nn.Linear(16, 10))
print(model)
```

## Step 7: Optimization Cycle\nTraining for 200 epochs using Adam and Cross-Entropy Loss.

```python
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)
for epoch in range(200):
  model.train()
  optimizer.zero_grad()
  loss = criterion(model(t_X_train), t_y_train)
  loss.backward()
  optimizer.step()
  print(f"Final Epoch Loss: {loss.item():.4f}")
```

## Step 8: Generalization Audit\nEvaluating precision, recall, and confusion matrix on the test set.

```python
from sklearn.metrics import classification_report
model.eval()
with torch.no_grad():
  test_preds = torch.argmax(model(t_X_test), dim=1).numpy()
  print(classification_report(y_test, test_preds, digits=3))
```
